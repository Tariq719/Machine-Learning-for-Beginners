#### **Which Encoding Method Should We Use?**

Before encoding, we should examine the unique values in each categorical feature.

The choice of encoding depends on whether the categories have a natural order or not.

#### **Types of Categorical Data**

Categorical features are generally divided into two types:

> ##### **Ordinal Data**

Categories have a natural order.

Examples:

- Small → Medium → Large
- Low → Medium → High
- Poor → Average → Excellent

Ordinal data is usually encoded using **Label Encoding**.

---

> ##### **Nominal Data**

Categories have no natural order.

Examples:

- Male, Female
- Computer Science, Mathematics, Physics
- Red, Blue, Green

Nominal data is usually encoded using **One-Hot Encoding**.


### **What is Label Encoding**

Label Encoding converts each category into a unique numerical value.

It is generally used for **Ordinal Data**, where the categories have a natural order.

Examples:

- Low → 0
- Medium → 1
- High → 2

Although Label Encoding can also be applied to Nominal Data, it may introduce an artificial order between categories. Therefore, it is usually **not recommended** for nominal features such as Gender and Department.

````python
# Import LabelEncoder
from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder Object
label_encoder = LabelEncoder()
````

````python
# Apply Label Encoding to Gender
X["gender"] = label_encoder.fit_transform(X["gender"])
# Display the First Five Records
X.tail()
````

````python
# Display Original Categories
label_encoder.classes_
````

````python
# Display Unique Encoded Values

X["gender"].unique()
````

#### **Observation**

Label Encoding successfully converted the Gender feature into numerical values.

However, Gender is a **Nominal** feature because there is no natural order between Male and Female.

Therefore, Label Encoding is **not the preferred encoding method** for this feature.

However, Label Encoding is commonly used for binary categorical features such as Gender.

### **What is One-Hot Encoding**

One-Hot Encoding converts each category into a separate binary (0 or 1) column.

It is generally used for **Nominal Data**, where there is no natural order between categories.

Examples:

Department

- CS
- AI
- DS
- IT
- SE

becomes

- department_CS
- department_AI
- department_DS
- department_IT
- department_SE

Each row contains **1** for the corresponding category and **0** for all other categories.

````python
# Display Unique Department Values

X["department"].unique()
````

````python
# Apply One-Hot Encoding

X = pd.get_dummies(
    X,
    columns=["department"],
    dtype=int
)


# Create another copy for demonstration
# X_onehot = X.copy()

# X_onehot = pd.get_dummies(
#     X_onehot,
#     columns=["gender", "department"],
#     dtype=int
# )


# Display the First Five Records

# X.head()
````

One-Hot Encoding removes the artificial order that Label Encoding would introduce for nominal data.

| Situation                                 | Recommended Method                                                   |
| ----------------------------------------- | -------------------------------------------------------------------- |
| Ordinal categories                        | ✅ Label Encoding                                                     |

### **What is Feature Scaling**

Feature Scaling is the process of bringing numerical features to a similar scale.

In many datasets, different features have different ranges.

For example:

| Feature | Values |
|----------|---------|
| Age | 18 – 25 |
| Attendance Percentage | 60 – 100 |
| Tuition Fee | 15000 – 50,000 |
| Semester | 1 - 8 |

Because these features have very different ranges, some Machine Learning algorithms may give more importance to larger values.

Feature Scaling helps ensure that each feature contributes fairly during model training.

````python
# Display Summary Statistics

X.describe()
````

#### Common Feature Scaling Techniques

The two most commonly used feature scaling techniques are:

- StandardScaler
- MinMaxScaler

````python
# Import Feature Scaling Methods

from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
````

#### **What is StandardScaler**

StandardScaler standardizes numerical features so that they have:

- Mean = 0
- Standard Deviation = 1

It is commonly used for Machine Learning algorithms that are sensitive to the scale of the data.

````python
# Create StandardScaler Object

standard_scaler = StandardScaler()
````

````python
# Select Numerical Features

numerical_features = [
    "age",
    "semester",
    "attendance_percentage",
    "tuition_fee"
]
````

````python
# Apply Standard Scaling

X_standard = X.copy()

X_standard[numerical_features] = standard_scaler.fit_transform(
    X_standard[numerical_features]
)

X_standard.head()
````

````python
# Display Summary Statistics

X_standard.describe()
````

### **How StandardScaler Works**

StandardScaler performs two steps for every numerical feature.

#### **Step 1**

Subtract the **Mean** from every value.

```
Difference = Value − Mean
```

#### **Step 2**

Divide the result by the **Standard Deviation**.

```
Scaled Value = (Value − Mean) / Standard Deviation
```

As a result:

- Values **below the Mean** become **negative**.
- Values **equal to the Mean** become approximately **0**.
- Values **above the Mean** become **positive**.

StandardScaler does **not** change the order of the data. It only changes the scale.

#### **Example**


| Student | Age |
| ------- | --: |
| A       |  18 |
| B       |  20 |
| C       |  22 |
| D       |  24 |
| E       |  26 |

##### **Calculate Means**

- (18 + 20 + 22 + 24 + 26) / 5 = 22.
- So difference from the mean

| Age |   Difference |
| --: | -----------: |
|  18 | 18 − 22 = -4 |
|  20 | 20 − 22 = -2 |
|  22 |  22 − 22 = 0 |
|  24 | 24 − 22 = +2 |
|  26 | 26 − 22 = +4 |


##### **Now Devide by Standard Deviation**

Suppose Standard Deviation = **2.83**

| Age | Difference | Scaled Value |
| --: | ---------: | -----------: |
|  18 |         -4 |        -1.41 |
|  20 |         -2 |        -0.71 |
|  22 |          0 |         0.00 |
|  24 |          2 |         0.71 |
|  26 |          4 |         1.41 |

These are the values you see after StandardScaler.

StandardScaler learns the Mean and Standard Deviation for each numerical feature independently.

#### **Should We Scale Every Column?**

No.

Only numerical features should be scaled.

Categorical features such as Gender and Department (after encoding) are usually not scaled because their values represent categories rather than continuous measurements.

### **MinMaxScaler**

MinMaxScaler transforms numerical features so that all values fall within a fixed range.

By default, the range is:

- Minimum becomes 0.

- Maximum becomes 1.

- Every other value is scaled proportionally.

Unlike StandardScaler, MinMaxScaler does not produce negative values.


It is commonly used when Machine Learning algorithms require features to be within the same range.

````python
# Import MinMaxScaler

from sklearn.preprocessing import MinMaxScaler
````

````python
# Create MinMaxScaler Object

minmax_scaler = MinMaxScaler()
````

````python
# Select Numerical Features

numerical_features = [
    "age",
    "semester",
    "attendance_percentage",
    "tuition_fee"
]
````

````python
# Apply Min-Max Scaling

X_minmax = X.copy()

X_minmax[numerical_features] = minmax_scaler.fit_transform(
    X_minmax[numerical_features]
)

# Display the First Five Records

X_minmax.head()
````

### Observation

MinMaxScaler converts all numerical features to the range **0 to 1**.

- The smallest value becomes **0**.
- The largest value becomes **1**.
- All other values are scaled proportionally between **0** and **1**.

Unlike StandardScaler, MinMaxScaler does not produce negative values.

#### StandardScaler vs MinMaxScaler

| StandardScaler       | MinMaxScaler           |
| -------------------- | ---------------------- |
| Mean = 0             | Minimum = 0            |
| Std = 1              | Maximum = 1            |
| Can produce negative values | No negative values  |
| Better when data is approximately normally distributed | Better when a fixed range (0–1) is desired                    |
| **Center around mean** | **Preserves the relative position within the original range** |

### **Train-Test Split**

#### **What is Train-Test Split?**

Train-Test Split is the process of dividing a dataset into two parts: one for training the machine learning model and another for testing its performance on unseen data.

---

#### **Training Set**

The **training set** is the portion of the dataset used to train the machine learning model. The model learns the relationship between the input features and the target variable from this data.

---

#### **Testing Set**

The **testing set** is the portion of the dataset used to evaluate the trained model. It measures how well the model performs on new data that was not used during training.

---

#### **Why Do We Use Train-Test Split?**

- Evaluate the model on unseen data.
- Measure the model's performance.
- Reduce the risk of overfitting.
- Improve model generalization.

---

#### **Common Train-Test Split Ratios**

| Training Set | Testing Set |
|:------------:|:-----------:|
| 80% | 20% |
| 70% | 30% |
| 75% | 25% |
| 90% | 10% |

> **Note:** The **80:20** split is the most commonly used ratio in machine learning projects.

#### **Splitting the Dataset**

The `train_test_split()` function is used to divide the feature matrix (**X**) and target variable (**y**) into training and testing datasets.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

#### **Understanding the Parameters**

The `train_test_split()` function accepts several parameters to control how the dataset is divided.

| Parameter | Description |
|-----------|-------------|
| `X` | Feature matrix (independent variables). |
| `y` | Target variable (dependent variable). |
| `test_size` | Proportion of data used for testing (e.g., `0.20` = 20%). |
| `train_size` | Proportion of data used for training. If omitted, it is calculated automatically. |
| `random_state` | Controls the random split. Using the same value produces the same train-test split every time. |
| `shuffle` | Randomly shuffles the data before splitting. Default is `True`. |
| `stratify` | Maintains the same class distribution in both training and testing datasets. Mainly used for classification problems. |
