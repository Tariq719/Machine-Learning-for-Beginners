##### **6.1 ndim Attribute**
The `ndim` attribute returns the number of dimensions of an array.

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks)
print("Dimension of Array is = ",student_marks.ndim)
```

```python
student_marks = np.array([[85, 78, 92], [88, 95, 81]]) 
print(student_marks)
print("Dimension of Array is = ",student_marks.ndim) 
```

##### **6.2 shape Attribute**
The `shape` attribute returns the size of an array along each dimension.

For a 2-D array, it is represented as: `(rows, columns)`

```python
student_marks = np.array([[85, 78, 92,20,40], [88, 95, 81,40,8],[88, 95, 81,40,50]]) 
student_marks.shape
```

##### **6.3 size Attribute**
The `size` attribute returns the total number of elements in an array. 

```python
student_marks = np.array([[85, 78, 92,20,40], [88, 95, 81,40,50],[88, 95, 81,40,50]]) 
student_marks.size
```

##### **6.4 dtype Attribute**
The `dtype` attribute returns the data type of the elements stored in an array. 

```python
student_marks = np.array([[85, 78, 92,20,40], [88, 95, 81,40,50],[88, 95, 81,40,50]]) 
student_marks.dtype
```

#### **7. Indexing & Slicing**
Each element in NumPy array has an index. `The first element has an index of 0` While `The second element has an index of 1, and so on.`

##### **7.1 Indexing 1-D Array**

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks[0]) 
print(student_marks[2]) 
print(student_marks[4]) 
```

> **Negative indexing**

Negative indexing starts from the end of the array.
- `-1` refers to the last element,
- `-2` refers to the second last element
- `-3` refers to the third last element.  

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks[-1]) 
print(student_marks[-2])
```

##### **7.2 slicing 1-D Array**
Slicing is used to retrieve a range of elements from an array.

`array[start:stop:step]`

```python
student_marks = np.array([85, 78, 92, 88, 95])

# Here we will give two index, first one is include while 2nd one is exclude
# Means if [3:5], it means start from 3 till to 4, means 5 is exclude
print(student_marks[3:5])
```

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks[2:]) 
```

```python
numbers = np.array([10, 20, 30, 40, 50, 60])

# Here the 3rd value show us step
print(numbers[0:6:2]) 
```

##### **7.3 Indexing 2-D arrays**
`2-D array consists of rows and columns. To access an element, specify the row index followed by the column index.`

```python
student_marks = np.array([[85, 78, 92],[88, 95, 81]]) 
print(student_marks[0,1]) 
print(student_marks[1,2])
```

##### **7.4 Slicing 2-D arrays**

```python
student_marks = np.array([[85, 78, 92],[88, 95, 81]]) 
print(student_marks[:,1:]) 
```

```python
student_marks = np.array([[85, 78, 92,98],[88, 95, 81,90],[1,2,3,4],[5,6,7,8]]) 
print(student_marks)
```

```python
print(student_marks[1:,2:]) 
```

##### **7.5 Boolean Indexing and Filtering**

```python
marks = np.array([45, 67, 82, 39, 91])

passed_marks = marks[marks >= 50]

print(passed_marks)
```

##### **8. Changing Data Types**

#### **10. Array Operation**
`Addition`, `Subtraction`, `Multiplication`, `Division`

```python
# Addition
sales_january = np.array([12000, 15000, 18000]) 
sales_february = np.array([10000, 16000, 17000]) 
total_sales = sales_january + sales_february 
print(total_sales)
```

```python
# Subtraction
planned_sales = np.array([15000, 18000, 20000]) 
actual_sales = np.array([14000, 17000, 19500]) 
difference = planned_sales - actual_sales 
print(difference) 
```

```python
# Multiplication
quantity = np.array([5, 8, 10]) 
price = np.array([200, 150, 100]) 
total_amount = quantity * price 
print(total_amount) 
```

```python
# Division
obtained_marks = np.array([420, 450, 470]) 
total_marks = np.array([500, 500, 500]) 
percentage = (obtained_marks / total_marks) * 100 
print(percentage) 
```

#### **11. Broadcasting**

Broadcasting allows NumPy to perform operations between arrays and compatible values or shapes without explicitly repeating the data.

```python
marks = np.array([70, 75, 80])

bonus = 5

updated_marks = marks + bonus

print(updated_marks)
```

#### **12. Scalar operations**
`A scalar is a single value. NumPy allows you to perform operations between an array and a single 
value.`

```python
# Add a Scalar value
marks = np.array([70, 75, 80]) 
updated_marks = marks + 5 
print(updated_marks) 
```

```python
# Multiply a scalar
salary = np.array([30000, 35000, 40000]) 
updated_salary = salary * 1.10 
print(updated_salary) 
```

```python
# Divide by a scalar value
salary = np.array([30000, 35000, 40000]) 
updated_salary = salary / 5
print(updated_salary) 
```

```python
# Subtract a scalar value
salary = np.array([30000, 35000, 40000]) 
updated_salary = salary - 5000
print(updated_salary) 
```

```python
# Practical Example
obtained_marks = np.array([425, 450, 475]) 
total_marks = 500 
percentage = (obtained_marks / total_marks) * 100 
print("Obtained Marks:", obtained_marks) 
print("Percentage:", percentage) 
```

#### **14. Working with Array Axes**

`axis=0 → operation down the rows / column-wise`

```python
marks = np.array([
    [80, 75, 90],
    [70, 85, 95],
    [88, 92, 84]
])

print(np.sum(marks))
print("Axis = 0")
print(np.sum(marks, axis=0))
```

`axis=1 → operation across the columns / row-wise`

```python
marks = np.array([
    [80, 75, 90],
    [70, 85, 95],
    [88, 92, 84]
])

print("Axis = 1")
print(np.sum(marks, axis=1))
```

#### **15.2 `flatten()` vs `ravel()`**

Both can convert a multi-dimensional array into a one-dimensional array.

`flatten()` returns a copy.

`ravel()` generally returns a view when possible.

> Flatten()

```python
# Create a numpy array
a = np.array([(1,2,3,4),(3,1,4,2)])

# Let's print the array a
print ("Original array:\n ", a)

print("===========================")

# Convert nd array to 1D array
c = a.flatten()

# Flatten passes copy of original array to 'c'
print ("\nFlatten array using flatten: ", c)
```

```python
# Changed the Value of the Copied Array but it will not make changes in the origonal Array
c[2] = 33

print ("Flatten array: ", c)

# Let's print the array a
print ("\nOriginal array:\n ", a)
```

> Ravel()

```python
import numpy as np

# Create a numpy array
a = np.array([(1,2,3,4),(3,1,4,2)])

# Let's print the array a
print ("Original array:\n ", a)

ra = np.ravel(a)
print ("\nFlatten array using ravel: ", ra)
```

```python
# Changed the Value of the Copied Array but it will also make changes in the origonal Array, as we use Ravel()
ra[3] = 44

print ("Flatten array: ", ra)

# Let's print the array a
print ("\nOriginal array:\n ", a)
```

#### **16. Random Numbers**
- rand()
- randint()
- seed()  

> rand() Function 
> * The rand() function generates random decimal numbers between 0 and 1.

```python
# 1-D Array
random_numbers = np.random.rand(10) 
print(random_numbers) 
```

```python
# 2-D Array
data = np.random.rand(2, 3)
print(data) 
```

> randint() Function 
> * The randint() function generates random integers within a specified range. 

```python
# np.random.randint(start, stop, size) 
marks = np.random.randint(50, 101, 10) 
print(marks) 
```

> seed() Function 

Normally, random numbers change every time a program is executed.

The seed() function fixes the starting point of the random number generator, allowing the same random numbers to be generated every time.

```python
np.random.seed(42) 
marks = np.random.randint(50, 101, 5) 
print(marks) 
```

```python
np.random.seed(43) 
marks = np.random.randint(50, 101, 5) 
print(marks) 
```

```python
# As i Use again but given same values as above
np.random.seed(42) 
marks = np.random.randint(50, 101, 5) 
print(marks)
```
