## Prerequisites

In the previous notebook, we learned how to:

- Create the feature matrix (`X`) and target variable (`y`)
- Encode categorical features
- Scale numerical features
- Split the dataset into training and testing sets

For this notebook, we will recreate these variables so that the notebook can be executed independently.

````python
# Import Required Libraries

import numpy as np
import pandas as pd

from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler

from sklearn.model_selection import train_test_split
````

````python
# Load the cleaned dataset
df = pd.read_csv("../datasets/student_performance_cleaned.csv")
````

````python
# Select the Target Variable

y = df["marks"]
````

````python
# Select Input Features
X = df.drop(columns=["marks",'full_name','student_id','email','city','admission_date','scholarship'])
````

````python
# Create OneHotEncoder Object
onehot_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

# Apply One-Hot Encoding
encoded_data = onehot_encoder.fit_transform(
    X[["department", "gender"]]
)

# Convert Encoded Data into DataFrame
encoded_df = pd.DataFrame(
    encoded_data,
    columns=onehot_encoder.get_feature_names_out(["department", "gender"]),
    index=X.index
)

# Remove Original Columns
X = X.drop(columns=["department", "gender"])

# Combine Encoded Columns with Remaining Features
X = pd.concat([X, encoded_df], axis=1)

# Display the Result
X.head()
````

````python
# Display Feature Names

X.columns
````

````python
# Display Summary Statistics

X.describe()
````

````python
# Create MinMaxScaler Object

minmax_scaler = MinMaxScaler()
````

````python
# Select Numerical Features

numerical_features = [
    "age",
    "semester",
    "attendance_percentage",
    "tuition_fee"
]
````

````python
# Apply Min-Max Scaling

X_minmax = X.copy()

X_minmax[numerical_features] = minmax_scaler.fit_transform(
    X_minmax[numerical_features]
)

# Display the First Five Records

X_minmax.head()
````

````python
# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_minmax,
    y,
    test_size=0.20,
    random_state=42
)
````

````python
# Display the shape of each dataset
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
````

````python
# Display the first five rows of the testing features
X_test.head()
````

````python
# Display the first five rows of the testing features
y_test.head()
````

````python
# Display the first five values of the testing target
X_train.head()
````

````python
# Display the first five values of the training target
y_train.head()
````

#### Train Your First Machine Learning Model

````python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
````

````python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = mse ** 0.5
r2 = r2_score(y_test, y_pred)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R² Score:", r2)
````

````python
# comparison table
results = pd.DataFrame({
    "Actual Marks": y_test,
    "Predicted Marks": y_pred
})

results.head(10)
````
