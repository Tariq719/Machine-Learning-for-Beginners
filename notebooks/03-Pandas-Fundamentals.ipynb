#### **4. Pandas DataFrame**

A DataFrame is a two-dimensional labeled data structure consisting of rows and columns.

It is the primary Pandas structure used for tabular data analysis.

#### **4.2 Creating a DataFrame from a NumPy Array**

##### **6.4 loc[] vs iloc[]**

| Method | Selection Method | Example |
|---|---|---|
| `loc[]` | Label-based | `df.loc[0, "marks"]` |
| `iloc[]` | Position-based | `df.iloc[0, 2]` |

##### Important Difference

- `loc[]` uses row/column labels.
- `iloc[]` uses integer positions.
- `loc[]` includes the ending label in a slice.
- `iloc[]` follows standard Python slicing and excludes the ending position.

##### **Selecting Data Using loc[]**

```python
# Select Single Row
df.loc[5]
```

```python
# Select Multiple Rows
df.loc[5:10]
```

```python
# Select Multiple Rows and Specific Columns
# df.loc[0:5,"marks"]
df.loc[0:5, ["full_name", "marks"]] 
```

##### **Selecting Data Using iloc[]**

#### **7. Filtering and Sorting Data**

##### **7.1 Filtering**

##### **Filtering with a Single Condition**

```python
# Select students whose Attendance Percentage are greater than 80.
df[df["attendance_percentage"] > 80]
```

```python
 # Select Students who study in CS Department 
df[df["department"] == "CS"]
```

```python
# Select Students who belong to Bannu City
df[df["city"] == "Bannu"]
```

##### **Filtering and Selecting Specific Columns**
After filtering rows, we can select only the columns we need from the resulting DataFrame.

```python
df[df["attendance_percentage"] > 92][["full_name", "marks","attendance_percentage"]]
```

```python
df.loc[df["attendance_percentage"] > 92, ["full_name", "marks","attendance_percentage"]]
```

##### **Filtering with Multiple Conditions**

##### **Filtering Operators**

- `&` means AND
- `|` means OR
- `~` means NOT
- Each condition should normally be enclosed in parentheses.

```python
# CS Students who belong to Bannu City
df[(df["department"] == "CS") & (df["city"] == "Bannu")]
```

```python
# Students from CS or AI 
df[(df["department"] == "CS") | (df["department"] == "AI")] 
```

```python
# Students Without Scholarships 
df[df["scholarship"] != "Yes"] 
# or 
# df[~(df["scholarship"] == "Yes")] 
```

##### **isin()**

`isin()` is useful when checking whether a column value belongs to a list of possible values.

```python
df[df["department"].isin(["DS", "SE"])]
```

##### **7.2 Sorting**
The `sort_values()` method sorts rows according to one or more columns.

- `ascending=True` sorts from smallest to largest or A to Z.
- `ascending=False` sorts from largest to smallest or Z to A.

```python
# Sort semester in Ascending Order 
df.sort_values(by="semester")
```

```python
# Sort Semester in Descending Order 
df.sort_values(by="semester", ascending=False)
```

```python
# Sort by Department and Semester 
df.sort_values(by=["department", "semester"], ascending=[True, False]) 
```

#### **8. Rename Columns**

```python
# Rename the marks column to Obtained Marks. 
df.rename(columns={"marks":"Obt_marks"})
```

```python
# Renaming multiple columns [full_name --> student_name] and [attendance_percentage --> attendance]
df.rename(columns={ 
"full_name":"student_name", 
"attendance_percentage":"attendance" 
}) 
```

##### **Checking Whether the Original DataFrame Changed**

When `rename()` is used without `inplace=True`, the original DataFrame is not modified.

The returned DataFrame contains the renamed columns.

```python
df.head()
```

##### **Creating a Modified DataFrame**

Instead of modifying the original object in place, we can assign the result back to `df`.

This makes the transformation explicit and easier to track.

```python
# To permanently save the changes in the existing DataFrame, use the inplace=True parameter.
# df.rename(columns={"marks":"obt_marks"}, inplace=True)

df_renamed = df.rename(columns={"marks": "obt_marks"})

df_renamed.head()
```

#### **9. Adding and Modifying Columns**

A DataFrame allows us to create new columns from existing data.

##### **Converting a Column to Numeric Data**
As here we will use `marks` column for different purposes, but as its data type is str, so we will first convert it into numeric.

The `to_numeric()` function converts values into numeric data types.

Using `errors="coerce"` converts values that cannot be interpreted as numbers into `NaN`.

This is useful when numeric data contains invalid or non-numeric values.

```python
# As we will use marks to use modification, so first we will convert to 'numeric', as it is 'str'
df['marks'] = pd.to_numeric(df['marks'], errors='coerce')
```

```python
df["bonus_marks"] = 5

df["final_marks"] = df["marks"] + df["bonus_marks"]

df[["full_name", "marks", "bonus_marks", "final_marks"]]
```

#### **10. Basic Data Analysis**

##### **10.1 df.count()**
`Purpose:` Counts the total number of values.

`Behavior:` It excludes empty or missing values (NaN).

```python
df["marks"].count()
```

##### **10.2 df.sum()**
`Purpose:` Adds all the numbers together.

`Behavior:` It calculates the grand total of the column.

```python
df["tuition_fee"].sum()
```

##### **10.3 df.mean()**
`Purpose:` Calculates the average score.

`Behavior:` It divides the sum by the count of values.

```python
df['marks'].mean()
```

##### **10.4 df.median()**

`median()` returns the middle value when the observations are ordered.

If there is an even number of observations, it returns the average of the two middle values.

`Purpose:` Finds the exact middle value.

`Behavior:` It sorts the numbers from lowest to highest and picks the center number.

```python
df['marks'].median()
```

##### **10.5 df.min()**
`Purpose:  ` Finds the lowest value.

`Behavior:` It searches for the minimum score in the column.

```python
df['marks'].min()
```

##### **10.6 df.max()**
`Purpose:` Finds the highest value.

`Behavior:` It searches for the maximum score in the column.

```python
df['marks'].max()
```

#### **12. Grouping and Aggregation**

The `groupby()` method allows us to divide data into groups and calculate statistics for each group.

```python
df.groupby("city")["marks"].mean()
```

```python
df.groupby("city")["marks"].agg(["count", "mean", "min", "max"])
```

```python
df.groupby("city").agg(
    Average_Marks=("marks", "mean"),
    Average_Attendance=("attendance_percentage", "mean")
)
```

```python
# Convert all full_name to Lower Case
df["full_name"].str.lower()
```

Let's combine the Pandas concepts learned in this notebook to answer a few questions about student performance.
