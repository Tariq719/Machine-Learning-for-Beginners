> loc[]
* Uses row labels
* Uses column names
* Label-based indexing
* includes both the start and end labels

> iloc[]
* Uses row positions
* Uses column positions
* Integer-based indexing
* follows Python slicing rules and 
excludes the ending position

#### Filter and Sorting Data

> Filtering

```python
# Students with Marks Greater Than 80 
# df[df["marks"] > 80]
df[df["city"]== "Bannu"]
```

```python
 # Students from the CS Department 
df[df["department"] == "CS"]
```

```python
# if required specific columns in filter then use

# df[df["attendance_percentage"] > 92][["full_name", "marks","attendance_percentage"]]

df.loc[df["attendance_percentage"] > 92, ["full_name", "marks","attendance_percentage"]]
```

> Filtering with Multiple Conditions 

```python
# CS Students who belong to Bannu City 
df[(df["department"] == "CS") & (df["city"] == "Bannu")]
```

```python
# Students from CS or AI 
df[(df["department"] == "CS") | (df["department"] == "AI")] 
```

```python
# Students Without Scholarships 
df[df["scholarship"] != "Yes"] 
# or 
# df[~(df["scholarship"] == "Yes")] 
```

> Sorting

```python
# Sort Marks in Ascending Order 
df.sort_values(by="semester")
```

```python
# Sort Marks in Descending Order 
df.sort_values(by="semester", ascending=False)
```

```python
# Sort by Department and Marks 
df.sort_values(by=["department", "semester"], ascending=[True, False]) 
```

#### Rename Columns

```python
# Rename the marks column to final_marks. 
df.rename(columns={"marks":"final_marks"})
```

```python
# Renaming multiple columns 
df.rename(columns={ 
"full_name":"student_name", 
"attendance_percentage":"attendance" 
}) 
```

```python
# check the df is parmanently changed or not
df
```

> Permanently renaming columns 

```python
# To permanently save the changes in the existing DataFrame, use the inplace=True parameter.
df.rename(columns={"marks":"final_marks"}, inplace=True)
```
