### Array Attributes
> * ndim
> * shape
> * size
> * dtype  

##### ndim Attribute
> The ndim attribute returns the number of dimensions of an array.

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks)
print("Dimension of Array is = ",student_marks.ndim)
```

```python
student_marks = np.array([[85, 78, 92], [88, 95, 81]]) 
print(student_marks)
print("Dimension of Array is = ",student_marks.ndim) 
```

##### shape Attribute
> The shape attribute returns the number of rows and columns in an array. 

```python
student_marks = np.array([[85, 78, 92,20,40], [88, 95, 81,40,8],[88, 95, 81,40,50]]) 
student_marks.shape
```

##### size Attribute
> The size attribute returns the total number of elements in an array. 

```python
student_marks = np.array([[85, 78, 92,20,40], [88, 95, 81,40,50],[88, 95, 81,40,50]]) 
student_marks.size
```

##### dtype Attribute 
> The dtype attribute returns the data type of the elements stored in an array. 

```python
student_marks = np.array([[85, 78, 92,20,40], [88, 95, 81,40,50],[88, 95, 81,40,50]]) 
student_marks.dtype
```

##### Indexing and Slicing
Each element in NumPy array has an index.
* The first element has an index of 0
* The second element has an index of 1, and so on. 

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks[0]) 
print(student_marks[2]) 
print(student_marks[4]) 
```

> Negative indexing 
> > Negative indexing starts from the end of the array.
> * -1 refers to the last element,
> * -2 refers to the second last element
> * and -3 refers to the third last element.  

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks[-1]) 
print(student_marks[-2])
```

##### Array slicing 
Slicing is used to retrieve a range of elements from an array.

##### Slicing 1-D Array

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks[:3])
```

```python
student_marks = np.array([85, 78, 92, 88, 95]) 
print(student_marks[2:]) 
```

```python
numbers = np.array([10, 20, 30, 40, 50, 60]) 
print(numbers[0:6:2]) 
```

##### 2-D Array

> Indexing 2-D arrays
* 2-D array consists of rows and columns. To access an element, specify the row index followed by the column index. 

```python
student_marks = np.array([[85, 78, 92],[88, 95, 81]]) 
print(student_marks[0,1]) 
print(student_marks[1,2]) 
```

> Slicing 2-D arrays
* Slicing can also be applied to 2-D arrays. 

```python
student_marks = np.array([[85, 78, 92],[88, 95, 81]]) 
print(student_marks[:,1:]) 
```

```python
student_marks = np.array([[85, 78, 92,98],[88, 95, 81,90],[1,2,3,4],[5,6,7,8]]) 
print(student_marks)
```

```python
print(student_marks[1:,2:]) 
```

##### Array Operation
* Addition
* Subtraction
* Multiplication
* Division

```python
# Addition
sales_january = np.array([12000, 15000, 18000]) 
sales_february = np.array([10000, 16000, 17000]) 
total_sales = sales_january + sales_february 
print(total_sales)
```

```python
# Subtraction
planned_sales = np.array([15000, 18000, 20000]) 
actual_sales = np.array([14000, 17000, 19500]) 
difference = planned_sales - actual_sales 
print(difference) 
```

```python
# Multiplication
quantity = np.array([5, 8, 10]) 
price = np.array([200, 150, 100]) 
total_amount = quantity * price 
print(total_amount) 
```

```python
# Division
obtained_marks = np.array([420, 450, 470]) 
total_marks = np.array([500, 500, 500]) 
percentage = (obtained_marks / total_marks) * 100 
print(percentage) 
```

##### Scalar operations 
> A scalar is a single value. NumPy allows you to perform operations between an array and a single 
value.

```python
# Adding Scalar
marks = np.array([70, 75, 80]) 
updated_marks = marks + 5 
print(updated_marks) 
```

```python
# Multiplying by a scalar
salary = np.array([30000, 35000, 40000]) 
updated_salary = salary * 1.10 
print(updated_salary) 
```

```python
# Divide by a scalar
salary = np.array([30000, 35000, 40000]) 
updated_salary = salary / 5
print(updated_salary) 
```

```python
# Subtract a scalar
salary = np.array([30000, 35000, 40000]) 
updated_salary = salary - 5000
print(updated_salary) 
```

```python
# Practical Example
obtained_marks = np.array([425, 450, 475]) 
total_marks = 500 
percentage = (obtained_marks / total_marks) * 100 
print("Obtained Marks:", obtained_marks) 
print("Percentage:", percentage) 
```

> std() Function 
> * The std() function calculates the standard deviation, which measures how much the values vary from the average.
> * A small standard deviation indicates that the values are close to the mean
> * while a large standard deviation indicates greater variation. 

```python
marks = np.array([85, 78, 92, 88, 95,80,82,75,70,72])
print(np.std(marks)) 
```

> var() Function 
> * The var() function calculates the variance of a dataset.
> * Variance measures how far the values are spread from the mean. It is closely related to standard deviation.
> * square root of VAR is equal to STD
> * Power 2 of STD is variance

```python
marks = np.array([85, 78, 92, 88, 95,80,82,75,70,72])
var1= np.var(marks)
print(var1) 
```

```python
marks = np.array([85, 78, 92, 88, 95,65,55,50,70,60])
var2 = np.var(marks)
print(var2) 
```

##### Reshaping Array
> * reshape()  
> * flatten()  

> The reshape() function changes the dimensions of an array while keeping the same data. The total number of elements must remain the same before and after reshaping. 

```python
numbers = np.array([10, 20, 30, 40, 50, 60]) 
new_array = numbers.reshape(2, 3) 
print(new_array) 
```

```python
numbers = np.array([10, 20, 30, 40, 50, 60])
print("1-D Array is = ",numbers)
new_array = numbers.reshape(3, 2) 
print(new_array) 
```

> flatten() Function
> * The flatten() function converts a multi-dimensional array into a one-dimensional array. 

```python
student_marks = np.array([
[85, 78, 92], 
[88, 95, 81] 
]) 
new_array = student_marks.flatten()
print("Old Array is")
print(student_marks)

print("----------")

print("New Array is")
print(new_array) 
```

##### Random Numbers
> * rand()
> * randint()
> * seed()  

> rand() Function 
> * The rand() function generates random decimal numbers between 0 and 1.

```python
# 1-D Array
random_numbers = np.random.rand(10) 
print(random_numbers) 
```

```python
# 2-D Array
data = np.random.rand(2, 3) 
print(data) 
```

> randint() Function 
> * The randint() function generates random integers within a specified range. 

```python
# np.random.randint(start, stop, size) 
marks = np.random.randint(50, 101, 10) 
print(marks) 
```

> seed() Function 
> * Normally, random numbers change every time a program is executed.
> * The seed() function fixes the starting point of the random number generator, allowing the same random numbers to be generated every time.

```python
np.random.seed(1) 
marks = np.random.randint(50, 101, 5) 
print(marks) 
```

```python
np.random.seed(2) 
marks = np.random.randint(50, 101, 5) 
print(marks) 
```
