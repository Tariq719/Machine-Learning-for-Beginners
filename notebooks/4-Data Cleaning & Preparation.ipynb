### Filling Methods

| Method                               | Best Used For                     | Example Columns                             | Why?                                                                                       |
| ------------------------------------ | --------------------------------- | ------------------------------------------- | ------------------------------------------------------------------------------------------ |
| **Mean (`mean()`)**                  | Continuous numeric data           | Marks, Salary, Height, Weight, Attendance % | Uses the average value. Suitable when there are no significant outliers.                   |
| **Median (`median()`)**              | Numeric data with outliers        | Age, Income, Tuition Fee                    | The median is not affected by extreme values, so it is more reliable when outliers exist.  |
| **Mode (`mode()`)**                  | Categorical/Text data             | Gender, City, Department, Scholarship       | Replaces missing values with the most frequently occurring value.                          |
| **Constant Value**                   | Text or business-specific data    | Email, Remarks, Status                      | Replace missing values with values like `"Unknown"` or `"Not Available"` when appropriate. |
| **Forward Fill (`ffill()`)**         | Sequential or time-series data    | Stock Price, Temperature, Sensor Data       | Copies the previous valid value to fill the missing value.                                 |
| **Backward Fill (`bfill()`)**        | Sequential or time-series data    | Stock Price, Temperature, Sensor Data       | Copies the next valid value to fill the missing value.                                     |
| **Drop Missing Values (`dropna()`)** | Rows with very few missing values | Any column                                  | Removes rows or columns containing missing values when it is safe to lose that data.       |


> How to use

| Data Type         | Mean Result | What to Do?                                                 |
| ----------------- | ----------- | ----------------------------------------------------------- |
| `float64`         | Decimal     | Fill directly with `mean()`                                 |
| `int64` / `Int64` | Decimal     | Round or convert to `int` before filling                    |
| `object`          | Text        | Use `mode()` or a constant value                            |
| `datetime64[ns]`  | Date        | Use a business rule, `ffill()`, `bfill()`, or leave missing |


> Fill Missing Values and convert to int where needed

````python
# df["gender"] = df["gender"].fillna("male")
df["gender"] = df["gender"].fillna(df["gender"].mode()[0])

# Fill with Mean
df["marks"] = df["marks"].fillna(round(df["marks"].mean())).astype(int)

# Fill with Median
df["age"] = df["age"].fillna(round(df["age"].median())).astype(int)

# Fill with Mode
df["city"] = df["city"].fillna(df["city"].mode()[0])

# Fill with Mean()
df["attendance_percentage"] = df["attendance_percentage"].fillna(df["attendance_percentage"].mean())

# Fill email with you choice text
df["email"] = df["email"].fillna("Not Available")

# Fill admission_date with any specific Date
# df["admission_date"] = df["admission_date"].fillna(pd.Timestamp("2024-01-01"))
df["admission_date"] = df["admission_date"].ffill()
# df["admission_date"] = df["admission_date"].bfill()

# Fill with mode()
df["scholarship"] = df["scholarship"].fillna(df["scholarship"].mode()[0])

# use mean() if no outlier and if outlier then use median()
# df["tuition_fee"] = df["tuition_fee"].fillna(round(df["tuition_fee"].mean()))
df["tuition_fee"] = df["tuition_fee"].fillna(round(df["tuition_fee"].median()))
````

#### **Outliers**
*An outlier is an unusually large or unusually small value that does not follow the general pattern of the data.*

***why do outlier occure***

| Cause               | Example                                        |
| ------------------- | ---------------------------------------------- |
| Data Entry Error    | Marks entered as **250** instead of **25**     |
| Measurement Error   | Faulty sensor records an incorrect temperature |
| Human Error         | Age entered as **150** instead of **15**       |
| Genuine Observation | A CEO earning much more than other employees   |


***How can we Handle Outlier***

| Method         | When to Use                                                    |
| -------------- | -------------------------------------------------------------- |
| Keep           | When the value is genuine and meaningful                       |
| Remove         | When it is an incorrect or impossible value                    |
| Correct        | When the correct value is known                                |
| Cap (`clip()`) | When you want to limit extreme values without removing records |


> How to Find Outlier

**Detecting Outliers Using Box Plot**

*A box plot is the most common visualization for detecting outliers.*

##### Detecting Outliers Using the IQR Method

````python
#  Interquartile Range (IQR) --> measures the spread of the middle 50% of the data.
# Step-1 Calculate Quartiles
Q1 = df["marks"].quantile(0.25) 
Q3 = df["marks"].quantile(0.75)

# Step-2 Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1 
print(IQR) 
````

````python
# Step 3: Calculate the Lower and Upper Limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Step 4: Display the Limits
print(lower_limit)
print(upper_limit)
````

````python
# Step 5: Find the Outliers
outliers = df[
(df["marks"] < lower_limit) | 
(df["marks"] > upper_limit)
] 
outliers 
````

> **This is why Data Analysis has two different concepts**

| Method         | Question it Answers                  |
| -------------- | ------------------------------------ |
| Business Rules | Is this value valid?                 |
| IQR Method     | Is this value statistically unusual? |

**Business Rule:** Marks must be between 0 and 100.

**So the professional workflow is**

- Step 1: Check Business Rules

- Step 2: Detect Statistical Outliers (IQR/Box Plot)

- Step 3: Investigate Each Outlier
- Decide:
- -  Keep
- - Correct
- - Remove
- - Cap

#### Decision 1: Invalid Values (Business Rule)

```text
Is the value impossible?
        │
      Yes
        │
Is the correct value known?
        │
   ┌────┴────┐
   │         │
 Yes        No
   │         │
Correct   Replace with NaN
             │
             ▼
       Fill Missing Values
```

**Example:**

- Marks = 250
- Marks = -15
- Age = 150
- Attendance = 120%

These values violate the business rules, so they should be corrected if possible. If the correct value is unknown, replace it with `NaN` and then handle the missing value using an appropriate method such as `mean()`, `median()`, or `mode()`.

#### Decision 2: Extreme Values (Statistical Outliers)

```text
Is the value possible?
        │
      Yes
        │
Is it extremely large or small?
        │
   ┌────┴───────────┐
   │                │
 Yes                No
   │                │
Machine Learning?   Keep It
        │
   ┌────┴────┐
   │         │
 Yes        No
   │         │
Clip()     Keep It
```

**Example:**

- CEO Salary = $2,000,000
- Luxury House Price = $10,000,000
- Customer Spending = $500,000

These values are **valid**, but they are much larger than most other observations. Instead of deleting them, data analysts may use **`clip()`** to reduce their influence when building statistical or machine learning models.

> **Remember:**

- **Invalid Values** violate business rules (e.g., Marks > 100, Age < 0). These should be corrected, replaced with `NaN`, or removed.
- **Outliers** are statistically unusual values. They are not necessarily incorrect and should be investigated before deciding to keep, remove, or cap (`clip()`) them.
