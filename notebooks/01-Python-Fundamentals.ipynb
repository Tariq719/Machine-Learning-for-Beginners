#### **5. Conditional Statements**

```python
# if condition

salary = int(input("Enter your salary = "))
exp = int(input("Enter your expenses = "))
if (salary > exp):
  print("Salary is more than expense")
```

```python
# if and else

obt = int(input("Enter your obtained marks = "))
total = int(input("Enter your total marks = "))
percent = obt/total *100
print(percent)
if (percent > 50):
  print("Excellent")
else:
  print("Good")
```

```python
# if --> elif --> else

obt = int(input("Enter your obtained marks = "))
total = int(input("Enter your total marks = "))
percent = obt/total *100
print(percent)
if (percent >= 90):
  print("A1")
elif (percent >= 80):
  print("A")
elif (percent >= 70):
  print("B")
elif (percent >=60):
  print("C")
elif (percent >= 50):
  print("D")
elif (percent >= 33):
  print("E")
else:
  print("Fail")
```

#### **10. File Handling**

`Reading data in a file Using (r)`

```python
# Used to read the existing file data
with open("../datasets/record.txt", "r") as f:
    print(f.read())
```

`Writing data in a file Using (w)`

```python
# Write New Sentences/Data inside a Text file, but if already present any text than it will be replaced
student_name = "Ali" 
marks = 85 
grade = "A" 
with open("../datasets/student_report.txt", "w") as file: 
    file.write("Student Report\n") 
    file.write("-----------------\n") 
    file.write("Name: " + student_name + "\n") 
    file.write("Marks: " + str(marks) + "\n") 
    file.write("Grade: " + grade)
```

```python
with open("../datasets/student_report.txt","r") as f:
    print(f.read())
```

```python
# Check here the previous record will be replaced
name = "Aisha" 
subject = "Computer" 
institute = "IT All Rounders" 
with open("../datasets/student_report.txt", "w") as file: 
    file.write("Marks Report\n") 
    file.write("-----------------\n") 
    file.write("Name: " + name + "\n") 
    file.write("subject: " + subject + "\n") 
    file.write("institute: " + institute + "\n \n")
```

```python
with open("../datasets/student_report.txt","r") as f:
    print(f.read())
```

`Writing data in a file Using (a) --> Append`

```python
# It will append the new data at the end of the file, and the existing data will be not erased
teacher_namee = "Salma" 
coursee = "Data Analysis" 
academye = "IT All Rounders" 
with open("../datasets/student_report.txt", "a") as file: 
    file.write("Teacher Report\n") 
    file.write("-----------------\n") 
    file.write("Teacher Name: " + teacher_namee + "\n") 
    file.write("Course: " + str(coursee) + "\n") 
    file.write("Academy: " + academye)
```

```python
with open("../datasets/student_report.txt","r") as f:
    print(f.read())
```
