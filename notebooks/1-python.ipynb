###Contidional Statments

```python
# if condition

salary = int(input("Enter your salary = "))
exp = int(input("Enter your expenses = "))
if (salary > exp):
  print("Salary is more then expense")
```

```python
# if and else

obt = int(input("Enter your obtained marks = "))
total = int(input("Enter your total marks = "))
percent = obt/total *100
print(percent)
if (percent > 50):
  print("Excellent")
else:
  print("Good")
```

```python
# if --> elif --> else

obt = int(input("Enter your obtained marks = "))
total = int(input("Enter your total marks = "))
percent = obt/total *100
print(percent)
if (percent >= 90):
  print("A1")
elif (percent >= 80):
  print("A")
elif (percent >= 70):
  print("B")
elif (percent >=60):
  print("C")
elif (percent >= 50):
  print("D")
elif (percent >= 33):
  print("E")
else:
  print("Fail")
```

File Handling

> Reading data in a file (r)

```python
with open("record.txt", "r") as f:
    print(f.read())
```

> Writing data in a file (w)

```python
student_name = "Ali" 
marks = 85 
grade = "A" 
with open("student_report.txt", "w") as file: 
    file.write("Student Report\n") 
    file.write("-----------------\n") 
    file.write("Name: " + student_name + "\n") 
    file.write("Marks: " + str(marks) + "\n") 
    file.write("Grade: " + grade)
```

```python
with open("student_report.txt","r") as f:
    print(f.read())
```

```python
name = "Aisha" 
subject = "Computer" 
institute = "IT All Rounders" 
with open("student_report.txt", "w") as file: 
    file.write("Marks Report\n") 
    file.write("-----------------\n") 
    file.write("Name: " + name + "\n") 
    file.write("subject: " + subject + "\n") 
    file.write("institute: " + institute + "\n \n")
```

```python
with open("student_report.txt","r") as f:
    print(f.read())
```
