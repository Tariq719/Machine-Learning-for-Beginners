#### **What is Machine Learning?**

Machine Learning is a branch of Artificial Intelligence (AI) that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed.

# 9.5 Features (X)

Features (`X`) are the independent variables or input variables used to train a machine learning model. They contain the information that helps the model learn patterns and make predictions.

**Example**

| Age | Attendance | Semester |
|----:|-----------:|---------:|
| 20 | 85 | 3 |
| 21 | 90 | 4 |
| 22 | 88 | 5 |

In this example, **Age**, **Attendance**, and **Semester** are the features (`X`).

# 9.6 Target Variable (y)

The target variable (`y`) is the dependent variable that the model tries to predict.

**Example**

| Age | Attendance | Semester | Marks |
|----:|-----------:|---------:|------:|
|20|85|3|78|
|21|90|4|86|
|22|88|5|81|

In this example, **Marks** is the target variable (`y`).

# 9.7 Training Data

Training data is the portion of the dataset used to train the machine learning model. The model learns the relationship between the features (`X`) and the target variable (`y`) using this data.

Typically, **70% to 80%** of the dataset is used for training.

# 9.8 Testing Data

Testing data is the portion of the dataset used to evaluate the trained model. It contains unseen records that help measure how well the model performs on new data.

Typically, **20% to 30%** of the dataset is used for testing.

# 9.9 Machine Learning Model

A machine learning model is an algorithm that learns patterns from the training data and uses those patterns to make predictions on new data.

Examples of machine learning models include:

- Linear Regression
- Logistic Regression
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)
- K-Nearest Neighbors (KNN)

# 9.10 Training a Model

Training is the process of teaching a machine learning model using the training dataset.

During training, the model learns the relationship between the features (`X_train`) and the target variable (`y_train`).

After training, the model is ready to make predictions on unseen data.

# 9.11 Prediction

Prediction is the process of using a trained machine learning model to estimate the target value for new or unseen data.

The `predict()` method is used to generate predictions.

```python
y_pred = model.predict(X_test)
```

# 9.12 Model Evaluation

Model evaluation measures how well a machine learning model performs by comparing the predicted values with the actual values.

Different evaluation metrics are used depending on the type of machine learning problem.

- **Regression:** MAE, MSE, RMSE, R² Score
- **Classification:** Accuracy, Precision, Recall, F1-Score

### 9.13 Regression vs Classification

Regression and Classification are the two main types of supervised machine learning problems. The choice depends on the type of target variable.

#### **Regression**

Regression is used to predict **continuous numerical values**.

**Examples:**

- Student Marks
- House Price
- Employee Salary
- Monthly Sales

**Common Algorithm**

- Linear Regression

---

#### **Classification**

Classification is used to predict **categories or class labels**.

**Examples:**

- Pass / Fail
- Yes / No
- Spam / Not Spam
- Placed / Not Placed

**Common Algorithm**

- Logistic Regression

---

### **Regression vs Classification**

| Regression | Classification |
|------------|----------------|
| Predicts numerical values | Predicts categories or classes |
| Continuous target variable | Categorical target variable |
| Example: Marks, Salary | Example: Pass/Fail, Yes/No |
| Uses Linear Regression | Uses Logistic Regression |

### **9.14 Linear Regression**

Linear Regression is a supervised machine learning algorithm used to predict **continuous numerical values** by learning the relationship between the features and the target variable.

Linear Regression should be used when the target variable is **continuous** and there is an approximately **linear relationship** between the features and the target.

### **Examples**

- Predict Student Marks
- Predict House Price
- Predict Employee Salary
- Predict Monthly Sales

### **Advantages**

- Simple and easy to understand.
- Fast to train.
- Easy to interpret.
- Works well for linear relationships.

### **Limitations**

- Suitable only for continuous target variables.
- Assumes a linear relationship between the features and the target.
- Sensitive to outliers.
- Cannot solve classification problems.

### **9.15 Logistic Regression**

Logistic Regression is a supervised machine learning algorithm used to predict **categorical or class labels** by estimating the probability that an observation belongs to a particular class.

Logistic Regression should be used when the target variable is **categorical** (e.g., Yes/No, Pass/Fail, Spam/Not Spam, Placed/Not Placed).

### **Examples**

- Predict Student Placement (Placed / Not Placed)
- Predict Pass or Fail
- Predict Spam or Not Spam
- Predict Disease (Positive / Negative)

### **Limitations**

- Assumes a linear relationship between the features and the log-odds of the target.
- Sensitive to outliers.
